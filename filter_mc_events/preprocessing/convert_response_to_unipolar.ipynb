```python
def to_unipolar(data):
    data_copy = {k : np.copy(v) for k, v in data.items()}
    data_copy['response'][5:,:,:] = 0
    data_copy['response'][:,5:,:] = 0
    return data_copy
```

```python
def to_unipolar_center_shape(data):
    data_copy = to_unipolar(data)
    for i in range(5):
        for j in range(5):
            data_copy['response'][i,j] = data_copy['response'][0,0]
    return data_copy
```

```python
to_unipolar(data);
```

```python
to_unipolar_center_shape(data);
```

```python
def trim_response_nticks(data, nticks=2500, velocity=0.16):
    data_copy = {k : np.copy(v) for k, v in data.items()}
    data_copy['response'] = data_copy['response'][:,:,nticks:]
    data_copy['drift_length'] = data_copy['drift_length'] - velocity * nticks * data_copy['time_tick']
    return data_copy
```

```python
trim_response_nticks(data)
```

```python
def get_label(res):
    return f'{res["drift_length"]:.3f}cm_{res["time_tick"]:.3f}us_bin_size{res["bin_size"]:.5f}'.replace('.', 'p')
```

```python
# unipolar, shape as center
res = to_unipolar_center_shape(data)
np.savez(f'response_v2a_center_unipolar_{get_label(res)}.npz', **res)
```

```python
res = trim_response_nticks(to_unipolar_center_shape(data))
np.savez(f'response_v2a_center_unipolar_{get_label(res)}.npz', **res)
```
