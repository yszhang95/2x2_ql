```python
with h5py.File("../threshold_summary.hdf5") as fthres:
    # print(fthres['io_group1/threshold'].dtype)
    for i in range(1,9):
        plt.figure()
        m = (fthres[f'io_group{i}/threshold']['Q'] < 25) & (fthres[f'io_group{i}/threshold']['Q'] > 2)
        y = fthres[f'io_group{i}/threshold']['y_i']/1000
        y = y[m]
        z = fthres[f'io_group{i}/threshold']['z_i']/1000
        z = z[m]
        Q = fthres[f'io_group{i}/threshold']['Q'][m]
        cx = plt.scatter(y, z, s=1, c=Q)
        plt.colorbar(cx)
```
